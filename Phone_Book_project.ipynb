```python
import pandas as pd
```

```python
phone_book={}

def add_contact():
    name=input("Enter the Name to add: ")
    print("Don't use char,+!@_etc... in number section")
    number=int(input("Enter the Number to add: "))

    phone_book[name]=number
    print("\ncontact added Successfully...")

def view_contact():
    if len(phone_book)==0:
        print("Empty contact list")
    else:
        print("-----Contacts-------")
        for name,number in phone_book.items():
            print("Name:",name)
            print("Mobile No:",number)
            print("--------------------\n")

def search_contact():
    name=input("Enter a Name search")
    if name in phone_book:
        print("Name:",name)
        print("phone:",phone_book[name])
    else:
        print("contact not found!..")

def delete_contact():
    name=input("Enter a Name delete")
    if name in phone_book:
        del phone_book[name]
        print("Contact deleted successfully")
    else:
        print("Contact not found!..")
        
        
print("=======Phone Book=========\n")
print("Choice:")
print("1, Adding contact")
print("2, View contact")
print("3, Searching contact")
print("4, Deleting contact")
print("0, Exit!!.")

while True:

    user_input = input("\nEnter your choice: ")
    try:
        choice = int(user_input)
    except ValueError:
        print("Invalid input! Please enter a number.")
        continue 

    if choice==1:
        add_contact()
        
    elif choice==2:
        view_contact()
        
    elif choice==3:
        search_contact()

    elif choice==4:
        delete_contact()
        
    elif choice == 0:
        print("Thank you for using the Phone Book...")
        break
    else:
        print("Invalid choise")   
```
